```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:\\Users\\HP\\Downloads\\Customerchurn.csv")
```

```python
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
print(df.describe())
```

```python
print(df["Churn"].value_counts())
print(df["Churn"].value_counts(normalize=True) * 100)
```

```python
print(df["gender"].value_counts())
```

```python
print(df["Contract"].value_counts())
```

```python
print(df["InternetService"].value_counts())
```

```python
print(df["PaymentMethod"].value_counts())
```

```python
print(df["MonthlyCharges"].describe())  # Monthly Charges 
```

```python
print("Total Revenue:", df["MonthlyCharges"].sum()) # Total Revenue
```

```python
print("Average Tenure:", df["tenure"].mean()) ## Average Tenure
```

```python
contract_churn = pd.crosstab(df["Contract"], df["Churn"])
print(contract_churn) ##Churn by Contract
```

```python
gender_churn = pd.crosstab(df["gender"], df["Churn"])
print(gender_churn)  # Churn by Gender
```

```python
internet_churn = pd.crosstab(df["InternetService"], df["Churn"])
print(internet_churn)  # Churn by Internet Service
```

```python
payment_churn = pd.crosstab(df["PaymentMethod"], df["Churn"])
print(payment_churn) # Churn by Payment Method
```

```python
#Histogram
plt.figure(figsize=(8,5))
plt.hist(df["MonthlyCharges"], bins=20)
plt.title("Monthly Charges Distribution")
plt.xlabel("Monthly Charges")
plt.ylabel("Customers")
plt.show()
```

```python
## Bar Chart – Churn
df["Churn"].value_counts().plot(kind="bar")

plt.title("Customer Churn")
plt.xlabel("Churn")
plt.ylabel("Count")
plt.show()
```

```python
#Bar Chart – Contract Type 
df["Contract"].value_counts().plot(kind="bar")

plt.title("Contract Type")
plt.xlabel("Contract")
plt.ylabel("Customers")
plt.show()
```

```python
#Pie Chart – Churn 
df["Churn"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Churn Percentage")
plt.show()
```
